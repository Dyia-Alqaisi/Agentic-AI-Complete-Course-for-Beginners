# Pydantic AI in Agentic AI

[Pydantic AI](https://pydantic.dev) is a production-grade Python framework and SDK built by the creators of [Pydantic](https://pydantic.dev). It is specifically designed to make building, type-checking, and managing AI agents reliable and safe for production environments.

## Core Capabilities

* **Strict Structured Outputs:** Large Language Models (LLMs) often generate unpredictable text. Pydantic AI uses Python type hints and standard Pydantic models to force LLMs into outputting strict, predictable JSON schemas.
* **Automated Error Correction:** If an LLM returns a response that fails validation, the framework automatically feeds the validation error back to the LLM, allowing it to correct its own output and retry.
* **Model Agnostic Framework:** It acts as a unified interface. You can switch between different model providers (like OpenAI, Anthropic, Gemini, DeepSeek, or Groq) by changing a simple configuration string.
* **Type-Safe Tool Integration:** It provides clean decorators to convert standard Python functions into tools that your agents can call, ensuring input arguments match expected types.
* **Dependency Injection:** It features a robust dependency injection system, making it easy to pass databases, API clients, or mock services into your agents and tools for seamless testing.

# How to Use Pydantic

### 1. Define a Pydantic model that represents the ideal schema of the data.
* This includes the expected fields, their types, and any validation constraints (e.g., `gt=0` for positive numbers).

### 2. Instantiate the model with raw input data (usually a dictionary or JSON-like structure).
* Pydantic will automatically **validate** the data and **coerce** it into the correct Python types (if possible).
* If the data doesn't meet the model's requirements, Pydantic raises a `ValidationError`.

### 3. Pass the validated model object to functions or use it throughout your codebase.
* This ensures that every part of your program works with **clean, type-safe, and logically valid data**.

# Annotated

### Direct Comparison

| Feature | First Block (`Annotated`) | Second Block (`= Field()`) |
| :--- | :--- | :--- |
| **Syntax Standard** | **Modern Pydantic v2 standard**. Uses Python's native `typing.Annotated` feature to attach metadata. | **Legacy syntax** inherited from Pydantic v1. Supported in v2 mostly for backward compatibility. |
| **Type Tooling Compatibility** | **Excellent**. IDEs and type checkers (Mypy, Pyright) see the raw type clearly, ensuring perfect autocomplete. | **Fair**. Type checkers can occasionally get confused because a `FieldInfo` object is being assigned to the variable. |
| **Code Reusability** | **High**. You can extract the `Annotated` type into a reusable type alias and share it across multiple models. | **Low**. You must copy-paste the `= Field(...)` configuration block every time you want to reuse it. |
| **Required Fields Clarity** | **Perfect**. Required fields have no assignment operator (`=`), making it explicitly clear to Python that no default exists. | **Ambiguous**. Required fields still use `= Field(...)`, meaning Python technically assigns an object instance as a placeholder default. |

---

### Key Structural Breakdown

#### 1. Why `Annotated` (First Block) is Recommended
Introduced natively in Python 3.9, `Annotated` lets you separate your core data type from its validation metadata. Pydantic extracts the `Field()` constraints under the hood without muddying the variable's type signature. 

This allows you to write highly maintainable code by creating reusable type aliases:
```python
# Defined once, reused anywhere
PatientAge = Annotated[int, Field(gt=0, description="Must be a positive integer")]

class PatientData(BaseModel):
    age: PatientAge
```

#### 2. The Direct Assignment Caveat (Second Block)
While assigning fields via `name: str = Field(...)` works perfectly fine at runtime, it leaves room for minor syntax bugs. For instance, in your second block's `email` and `linkedIn` lines, you use `= Field(...)` without specifying `default=None`. Pydantic correctly processes this as a required field, but external static code analyzers might flag it as having a non-standard default assignment.

# Pydantic V2: Validators, `cls` vs `self`, and Best Practices

## 1. Why do we write `@field_validator` and `@classmethod` together?
This is the standard syntax for custom field validation in Pydantic. Stacking these decorators tells the code two distinct things:
*   **`@classmethod` (Python's Job):** Explicitly tells Python that the method belongs to the class blueprint (`cls`), meaning it can be run before an object instance (`self`) actually exists.
*   **`@field_validator('field_name')` (Pydantic's Job):** Functions as a targeting system. It tells Pydantic to intercept incoming data for that specific field and route it through this function before building the final object.

---

## 2. 💡 The Pydantic V2 Pro-Tip
In **Pydantic V2**, the library automatically treats `@field_validator` methods as class methods under the hood. This means **`@classmethod` is now entirely optional**. 

You can write cleaner, shorter code like this:
```python
@field_validator('name')
def transform_name(cls, value):  # Still accepts 'cls', but no @classmethod needed!
    return value.title()
```

---

## 3. Quick Reference: `cls` vs `self` in Pydantic

| Concept | `cls` (Class / Blueprint Level) | `self` (Instance / Object Level) |
| :--- | :--- | :--- |
| **What it represents** | The class blueprint itself (`PatientData`). | A specific, fully-built object (`patient1`). |
| **When it runs** | **Before** the object is created. | **After** validation passes and the object is created. |
| **What it receives** | The **raw data values** (e.g., checking if `age = 60` or `email = "test@com"` is valid). | The **live attributes** stored inside the active object. |
| **Where to use it** | Inside `@field_validator` or `@model_validator` methods. | Inside regular model methods (like `def add_patient_data(self, ...)`). |

---

## 4. Visual Timeline of Data Flow

When you execute `patient = PatientData(name="john doe", age=60)`:

```text
 1. RAW DATA INPUT  --->   2. VALIDATION (Uses 'cls')   --->   3. OBJECT CREATION (Uses 'self')
   {"name": "john doe"}    Runs transform_name(cls, value)     Object is saved to 'patient'.
                           Returns "John Doe"                  You can now use self.name.

```


## 5.💡 Summary Checklist

```text 
If you are checking raw data before an object is made, use cls (Validators).
If you are reading or using data from an object that already exists, use self (Methods).
```

# Serialization in Pydantic V2

## 1. What is Serialization?
**Serialization** is the exact opposite of validation. While validation is the process of taking raw data *in* and turning it into a Python object, serialization is the process of exporting that completed Python object *out* into standard formats like **dictionaries** or **JSON strings**.

```text
VALIDATION:    Raw Data (Dict/JSON)  ───>  [Pydantic Checks]  ───>  Python Object (self)
SERIALIZATION: Python Object (self)   ───>  [Pydantic Export]  ───>  Clean Dict / JSON String
```

---

## 2. The Two Core Serialization Methods
Pydantic V2 provides two built-in, high-performance methods (powered by a Rust engine under the hood) to serialize your data:

### 1. `.model_dump()` ──> Outputs a Python Dictionary
Converts your model instance into a standard Python `dict`. This is ideal when passing data to databases, internal services, or standard Python logic.
```python
patient = PatientData(name="john doe", age=30, email="john@example.com")

patient_dict = patient.model_dump()
print(patient_dict)
# Output: {'name': 'John Doe', 'email': 'john@example.com', 'age': 30, 'weight': None, ...}
```

### 2. `.model_dump_json()` ──> Outputs a JSON String
Converts your model instance directly into a valid JSON string. This is ideal when sending data across a web network API or writing data straight into a text file.
```python
patient_json = patient.model_dump_json()
print(patient_json)
# Output: {"name":"John Doe","email":"john@example.com","age":30,"weight":null,...}
```

## 3. Power Filters: Shaping Your Exported Data
You can pass configuration rules straight inside `.model_dump()` or `.model_dump_json()` to cleanly shape what gets exported:

*   **`exclude_none=True`:** Strips out any fields completely if their value is `None`. Keeps your exports lean.
    ```python
    # Removes fields like 'weight' or 'married' if they are set to None
    clean_dict = patient.model_dump(exclude_none=True)
    ```
*   **`exclude_unset=True`:** Only exports fields that the user explicitly provided when creating the model, ignoring default values.
*   **`include` or `exclude` sets:** Allows you to cherry-pick or hide specific fields on the fly (perfect for hiding sensitive clinical data).
    ```python
    # Generates a dict that completely omits the email address
    public_dict = patient.model_dump(exclude={'email'})
    ```


````python
from pydantic import BaseModel, EmailStr, AnyUrl,Field,field_validator, model_validator,computed_field
from typing import  Optional,Annotated

class Address(BaseModel):
    street: str
    city: str
    state: str
    zip_code: str
    country: str


class PatientData(BaseModel):
    name :str
    email: EmailStr
    age:int
    weight: Optional[float] = None
    height: Optional[float] = None
    married: Optional[bool] = None
    allergies: Optional[list[str]] = None
    contact_info: Optional[dict[str, str]] = None
    address: Optional[Address] = None  # Optional field for address

    @model_validator(mode='after')
    def emergency_contact(self):
        if self.age >60 and 'emergency' not in self.contact_info:
            raise ValueError("Patients over 60 must have an emergency contact in their contact_info.")
        return self

    @field_validator('email')
    @classmethod
    def validate_email(cls, value):
        valid_domains = ['example.com', 'test.com']
        domain = value.split('@')[-1]
        
        if domain not in valid_domains:
            raise ValueError(f"Email domain must be one of {valid_domains}.")
        return value

    @field_validator('name')
    @classmethod
    def transform_name(cls, value):
        return value.title()  # Capitalize the first letter of each word in the name

    @computed_field
    @property
    def computed_bmi(self) -> float | None:
        if self.weight is not None and self.height is not None:
            return round(self.weight / (self.height ** 2), 2)
        return None




address_data = {
    "street": "123 Main St",
    "city": "Anytown",
    "state": "CA",
    "zip_code": "12345",
    "country": "USA"
}

data = {
    "name": "John Doe",
    "weight": 70.5,
    "age": 60, # This will raise a validation error because age should be an integer, by 
    "married": True,
    "allergies": ["penicillin", "shellfish"],
    "contact_info": {"phone": "123-456-7890", "email": "john.doe@example.com", "emergency": "Jane Doe"},
    "email": "john.doe@example.com",
    "linkedIn": "https://www.linkedin.com/in/johndoe",
    "height": 1.75,
    "address": address_data

}

patient_1= PatientData(**data) # ** is used to unpack the dictionary into keyword arguments for the model validation.
print('patient_1:', patient_1)

temp_dict = patient_1.model_dump()  # Convert the Pydantic model to a dictionary
print("\nConverted to dictionary:")
print(temp_dict)

temp_json = patient_1.model_dump_json()  # Convert the Pydantic model to a JSON string
print("\nConverted to JSON string:")
print(temp_json)
````
