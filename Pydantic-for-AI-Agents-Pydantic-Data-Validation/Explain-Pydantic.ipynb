# Pydantic AI in Agentic AI

[Pydantic AI](https://pydantic.dev) is a production-grade Python framework and SDK built by the creators of [Pydantic](https://pydantic.dev). It is specifically designed to make building, type-checking, and managing AI agents reliable and safe for production environments.

## Core Capabilities

* **Strict Structured Outputs:** Large Language Models (LLMs) often generate unpredictable text. Pydantic AI uses Python type hints and standard Pydantic models to force LLMs into outputting strict, predictable JSON schemas.
* **Automated Error Correction:** If an LLM returns a response that fails validation, the framework automatically feeds the validation error back to the LLM, allowing it to correct its own output and retry.
* **Model Agnostic Framework:** It acts as a unified interface. You can switch between different model providers (like OpenAI, Anthropic, Gemini, DeepSeek, or Groq) by changing a simple configuration string.
* **Type-Safe Tool Integration:** It provides clean decorators to convert standard Python functions into tools that your agents can call, ensuring input arguments match expected types.
* **Dependency Injection:** It features a robust dependency injection system, making it easy to pass databases, API clients, or mock services into your agents and tools for seamless testing.

# How to Use Pydantic

### 1. Define a Pydantic model that represents the ideal schema of the data.
* This includes the expected fields, their types, and any validation constraints (e.g., `gt=0` for positive numbers).

### 2. Instantiate the model with raw input data (usually a dictionary or JSON-like structure).
* Pydantic will automatically **validate** the data and **coerce** it into the correct Python types (if possible).
* If the data doesn't meet the model's requirements, Pydantic raises a `ValidationError`.

### 3. Pass the validated model object to functions or use it throughout your codebase.
* This ensures that every part of your program works with **clean, type-safe, and logically valid data**.


```python
from pydantic import BaseModel


class PatientData(BaseModel):
    name: str
    age: int
    weight: float = None  # Optional field with default value None


def add_patient_data(patient: PatientData):
        print(f"Adding patient data: Name={patient.name}, Age={patient.age}, Weight={patient.weight}")
        print("Patient data added successfully.")

def update_patient_data(patient: PatientData):
        print(f"Updating patient data: Name={patient.name}, Age={patient.age}, Weight={patient.weight}")
        print("Patient data updated successfully.")



data = {
    "name": "John Doe",
    "weight": 70.5,
    "age": '60' # This will raise a validation error because age should be an integer, by 
}

patient_1= PatientData(**data) # ** is used to unpack the dictionary into keyword arguments for the model validation.
add_patient_data(patient_1)
update_patient_data(patient_1)
```

# Little Complex Pydantic

```python
# 1. Native list vs typing.List
# First Block: Uses list[str]. Starting in Python 3.9 (via PEP 585), Python allowed
# developers to use standard collections (like list, dict, set) directly as generics
# for type hinting. This is the modern, recommended Pythonic approach
#
# Second Block: Uses List[str]. Before Python 3.9, standard types couldn't handle
# generic arguments like [str]. Developers had to import the capitalized List from the typing module.




from pydantic import BaseModel
from typing import List, Optional

class PatientData(BaseModel):
    name: str
    age: int
    weight: float = None  # Optional field with default value None
    married: bool = None  # Optional field with default value None
    allergies: list[str] = None  # Optional field with default value None
    contact_info: dict[str, str] = None  # Optional field with default value None


def add_patient_data(patient: PatientData):
        print(f"Adding patient data: Name={patient.name}, Age={patient.age}, Weight={patient.weight}, Married={patient.married}, Allergies={patient.allergies}, Contact Info={patient.contact_info}")
        print("Patient data added successfully.")

def update_patient_data(patient: PatientData):
        print(f"Updating patient data: Name={patient.name}, Age={patient.age}, Weight={patient.weight}, Married={patient.married}, Allergies={patient.allergies}, Contact Info={patient.contact_info}")
        print("Patient data updated successfully.")



data = {
    "name": "John Doe",
    "weight": 70.5,
    "age": 60, # This will raise a validation error because age should be an integer, by 
    "married": True,
    "allergies": ["penicillin", "shellfish"],
    "contact_info": {"phone": "123-456-7890", "email": "john.doe@example.com"}
}

patient_1= PatientData(**data) # ** is used to unpack the dictionary into keyword arguments for the model validation.
print('patient_1:')
add_patient_data(patient_1)
print("\nUpdating patient data...")
update_patient_data(patient_1)
```

# Data Validation 

```python
# 1. Native list vs typing.List
# First Block: Uses list[str]. Starting in Python 3.9 (via PEP 585), Python allowed
# developers to use standard collections (like list, dict, set) directly as generics
# for type hinting. This is the modern, recommended Pythonic approach
#
# Second Block: Uses List[str]. Before Python 3.9, standard types couldn't handle
# generic arguments like [str]. Developers had to import the capitalized List from the typing module.




from pydantic import BaseModel, EmailStr, AnyUrl
from typing import List, Optional



class PatientData(BaseModel):
    name: str
    age: int
    email: EmailStr # This field will validate that the input is a valid email address.
    linkedIn: AnyUrl # This field will validate that the input is a valid URL.
    weight: float = None  # Optional field with default value None
    married: Optional[bool] = None  # Optional field with default value None
    allergies:  Optional[list[str]] = None  # Optional field with default value None
    contact_info: Optional[dict[str, str]] = None  # Optional field with default value None


def add_patient_data(patient: PatientData):
        print(f"Adding patient data: Name={patient.name}, Age={patient.age}, Email={patient.email}, LinkedIn={patient.linkedIn}, Weight={patient.weight}, Married={patient.married}, Allergies={patient.allergies}, Contact Info={patient.contact_info}")
        print("Patient data added successfully.")

def update_patient_data(patient: PatientData):
        print(f"Updating patient data: Name={patient.name}, Age={patient.age}, Email={patient.email}, LinkedIn={patient.linkedIn}, Weight={patient.weight}, Married={patient.married}, Allergies={patient.allergies}, Contact Info={patient.contact_info}")
        print("Patient data updated successfully.")



data = {
    "name": "John Doe",
    "weight": 70.5,
    "age": 60, # This will raise a validation error because age should be an integer, by 
    "married": True,
    "allergies": ["penicillin", "shellfish"],
    "contact_info": {"phone": "123-456-7890", "email": "john.doe@example.com"}
}

patient_1= PatientData(**data) # ** is used to unpack the dictionary into keyword arguments for the model validation.
print('patient_1:')
add_patient_data(patient_1)

print("\nUpdating patient data...")
update_patient_data(patient_1)
```
