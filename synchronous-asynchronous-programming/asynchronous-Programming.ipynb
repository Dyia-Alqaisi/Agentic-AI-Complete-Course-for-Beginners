# Asynchronous Programming

```python
import asyncio
import time

async def fetch_weather():
    print("Fetching weather data...")
    await asyncio.sleep(1)  # Simulate a delay in fetching data
    print("Weather data fetched successfully.")


async def fetch_news():
    print("Fetching news data...")
    await asyncio.sleep(2)  # Simulate a delay in fetching data
    print("News data fetched successfully.")


async def main():
    start_time = time.time()

    # await fetch_weather() # Sequential Execution (Slower) 
    # await fetch_news() # Sequential Execution (Slower)

    await asyncio.gather(fetch_weather(), fetch_news())  # Concurrent Execution (Faster)

    end_time = time.time()
    print(f"Total time taken: {end_time - start_time:.2f} seconds")
```

```python
if __name__ == "__main__":
    await main()  # This line is not needed when using asyncio.run(main())
```
